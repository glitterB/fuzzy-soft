```python
#Importing dependencies
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Defining our function to minimize
def func(x):
    x1 = x[0]
    x2 = x[1]
    return (4*x1**2)+(3*x2**2)-(6*x1*x2)-(4*x1)

#Defining optimization parameters
lambdA = 1.2
eta = 0.075
N = 100
```

```python
#initial solution
x = np.array([0.0,0.0]).reshape(-1,1)

while lambdA>eta:
    f = func(x)

    for i in range(N):
        #Generating random numbers between -1 to 1
        random_nos = np.random.uniform(low=-1.0, high=1.0, size=(2, 1))
        u = random_nos/np.linalg.norm(random_nos)
    
        #New x
        x_new = x+(lambdA*u)
        #New function value
        f_new = func(x_new)
        if f_new<f:
            break
        x = x_new
        lambdA=lambdA/2
        
```
