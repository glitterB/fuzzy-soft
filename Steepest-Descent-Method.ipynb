```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import math as math
```

```python
# Step 1: Define the function and its derivative
def f(x):
    return x**2

def df(x):
    return 2*x
```

```python
# Step 2: Set hyperparameters
x = 10           # Initial guess
learning_rate = 0.1
iterations = 25
```

```python
# Step 3: Store history for plotting
x_history = [x]
f_history = [f(x)]
```

```python
# Step 4: Gradient descent loop
for i in range(iterations):
    grad = df(x)
    x = x - learning_rate * grad
    
    # Save values
    x_history.append(x)
    f_history.append(f(x))
```

```python
# Step 5: Plotting the results
plt.figure(figsize=(10, 4))

# Function value vs iteration
plt.subplot(1, 2, 1)
plt.plot(f_history, marker='o')
plt.title("Function Value over Iterations")
plt.xlabel("Iteration")
plt.ylabel("f(x)")

# Descent path on function curve
plt.subplot(1, 2, 2)
x_vals = [i / 100 for i in range(-1000, 1000)]
y_vals = [f(i) for i in x_vals]
plt.plot(x_vals, y_vals, label='f(x) = x^2')
plt.plot(x_history, f_history, 'ro-', label='Descent Path')
plt.title("Descent Path on Curve")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()

plt.tight_layout()
plt.show()
```
