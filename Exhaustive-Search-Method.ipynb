```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Let's define a function, whose maximum value is to be found
def f(x):
    return (2*x**2)-x**3
```

```python
#Defining a variables 
n = 100 #determines no of divisions
x = np.linspace(0,2,n)
```

```python
y = f(x)
```

```python
#let's plot our function
fig, ax = plt.subplots()
ax.plot(x,y)
```

```python
#let's run the maximum or mininum value finding algorithm
i = 2
while True:
    try:
        x_1 = x[i-1]
        x_2 = x[i]
        x_3 = x[i+1]
        if f(x_1)<f(x_2) and f(x_2)>f(x_3):
            print(x_2, f(x_2))
            break
        else:
            i+=2
    except IndexError:
        print('NOT FOUND')
        break
```
